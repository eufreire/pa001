```python
import random

tentativas = 6

palavrs = ['areio', 'areia', 'matas', 'tocos']

palavr = random.choice(palavrs)

acerto = []
for letra in palavr:
    acerto.append(acerto)

qtd_letras = len(acerto)

print(f'Tente adivinhar um nome com {qtd_letras} acerto!')

while tentativas > 0 :
    chut = input(f'\nVocê tem {tentativas} tentativas. Digite um nome com {qtd_letras} letras: ').lower()

    qtd_chuts = len(chut)

    while qtd_chuts != qtd_letras:
        chut = input(f'\nVocê tem {tentativas} tentativas. Digite um nome com {qtd_letras} letras: ').lower()
        qtd_chuts = len(chut)
    resultado = []


    for i in range(qtd_letras):
        if chut[i] == acerto[i]:

            resultado.append(chut[i].upper())

        elif chut[i] in acerto:

            resultado.append(chut[i])
        else:

            resultado.append('_')
    tentativas -=1

    print('Resultado: ', ' '.join(resultado))

    if chut == palavr:
        print('Parabéns! Você acertou a palavra!')

        break

    if tentativas == 0:
        print(f'Você perdeu! A palavra correta era: {palavr}')
```
