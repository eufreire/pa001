#Dicionario, Tuplas e Listas

```python
##Exercício 1: Manipulação de Listas

cidades = [ "são paulo", "são bernardo","maua", "riu grande","ribeirão pires" ]
cidades.append("são roque")

cidades.remove("são bernardo")
print(cidades)
```

```python
##Exercício 2: Operações com Tuplas

cores = ("Branco", "preto " ,"roxo")
mais_cores = cores + ('amrelo', 'azul')
print(mais_cores)
```

```python
##Exercício 3: Dicionários Simples

pessoa = {
    'nome': 'kaike',
    'idade': 19,
    'cursinho': 'canto'
}
pessoa['nota']= '9.4'
print(pessoa)
```

```python
##Exercício 4: Listas e Dicionários

venda = {
    'baike': 500,
    'celular': 1500,
    "bola":80}
venda["notebook"]=2000
venda["baike"]=5000
print(venda)
```

```python
##Exercício 5: Analisando uma Lista de Tuplas

pessoas = [("Marcos", 44), ("Bruno", 55), ("Maria", 60)]
pessoa_mais_velha = max(pessoas, key=lambda x: x[1])
print(pessoa_mais_velha)

media_idade = sum(pessoa[1] for pessoa in pessoas) / len(pessoas)
print(media_idade)
```
